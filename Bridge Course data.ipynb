````python
# =============================================================================
# INSTALLATION (Run this cell first in Colab)
# =============================================================================
# !apt-get install -y poppler-utils
# !pip install pdf2image g4f curl_cffi

import os
import io
import gc
import time
import g4f
from g4f.client import Client
from pdf2image import convert_from_path
from google.colab import files

# =============================================================================
# CONFIGURATION
# =============================================================================
MAX_RETRIES = 3  # Number of retry attempts per page per model
RETRY_DELAY = 2  # Initial delay between retries (seconds)
PAGE_DELAY = 1   # Delay between pages to avoid rate limiting

# Best OCR/Vision Models (in priority order)
OCR_MODELS = [

    {
        "provider": "PollinationsAi",
        "model": "gemini-search",
        "name": "claude40opusthinking-OCR",
        "priority": 1
    },
    {
        "provider": "DeepInfra",
        "model": "deepseek-ai/DeepSeek-OCR",
        "name": "DeepSeek-OCR",
        "priority": 6
    },
    {
        "provider": "DeepInfra",
        "model": "allenai/olmOCR-2-7B-1025",
        "name": "OLM-OCR",
        "priority": 2
    },
    {
        "provider": "DeepInfra",
        "model": "PaddlePaddle/PaddleOCR-VL-0.9B",
        "name": "PaddleOCR",
        "priority": 3
    },
    {
        "provider": "DeepInfra",
        "model": "microsoft/Phi-4-multimodal-instruct",
        "name": "Phi-4-Multimodal",
        "priority": 4
    },
    {
        "provider": "OperaAria",
        "model": "aria",
        "name": "Opera-Aria",
        "priority": 5
    }
]

# =============================================================================
# RETRY MECHANISM
# =============================================================================
def retry_with_backoff(func, max_retries=MAX_RETRIES, initial_delay=RETRY_DELAY):
    """
    Executes a function with exponential backoff retry logic.

    Args:
        func: Function to execute (should return the result)
        max_retries: Maximum number of retry attempts
        initial_delay: Initial delay in seconds (doubles each retry)

    Returns:
        Result from func() or raises the last exception
    """
    delay = initial_delay
    last_exception = None

    for attempt in range(max_retries):
        try:
            result = func()
            if attempt > 0:
                print(f"      ✓ Succeeded on attempt {attempt + 1}")
            return result
        except Exception as e:
            last_exception = e
            if attempt < max_retries - 1:
                print(f"      ⚠ Attempt {attempt + 1} failed: {str(e)[:100]}")
                print(f"      ⏳ Retrying in {delay}s...")
                time.sleep(delay)
                delay *= 2  # Exponential backoff
            else:
                print(f"      ✗ All {max_retries} attempts failed")

    raise last_exception


def clean_extracted_text(text):
    """
    Remove any unwanted headers or metadata that AI might add
    """
    lines = text.split('\n')
    cleaned_lines = []

    for line in lines:
        line_lower = line.lower().strip()

        # Skip common unwanted patterns
        if any([
            line_lower.startswith('page ') and line_lower[5:].strip().isdigit(),
            line_lower.startswith('extracted by:'),
            line_lower.startswith('extracted from:'),
            line_lower == '---' and len(cleaned_lines) < 3,  # Remove separator at start
            'here is the extracted' in line_lower,
            'the image contains' in line_lower,
            'ocr result' in line_lower,
        ]):
            continue

        cleaned_lines.append(line)

    # Remove leading empty lines
    while cleaned_lines and not cleaned_lines[0].strip():
        cleaned_lines.pop(0)

    return '\n'.join(cleaned_lines)


# =============================================================================
# MULTI-MODEL OCR WITH FALLBACK
# =============================================================================
def extract_page_with_models(image_path, client, models_list=OCR_MODELS, prompt=""):
    """
    Tries multiple models in sequence until one succeeds.

    Args:
        image_path: Path to the image file
        client: G4F client instance
        models_list: List of model configurations to try
        prompt: OCR extraction prompt

    Returns:
        Tuple of (extracted_text, successful_model_name)
    """
    for model_config in models_list:
        model_name = model_config["model"]
        friendly_name = model_config["name"]

        print(f"    🤖 Trying model: {friendly_name} ({model_name})")

        def try_model():
            with open(image_path, "rb") as image_file:
                response = client.chat.completions.create(
                    model=model_name,
                    messages=[
                        {
                            "role": "user",
                            "content": prompt
                        }
                    ],
                    image=image_file
                )

            extracted_text = response.choices[0].message.content

            if not extracted_text or len(extracted_text.strip()) < 10:
                raise ValueError("Model returned empty or very short response")

            return extracted_text

        try:
            # Try this model with retry mechanism
            extracted_text = retry_with_backoff(try_model, max_retries=MAX_RETRIES)
            print(f"    ✅ Success with {friendly_name}!")
            return extracted_text, friendly_name

        except Exception as e:
            print(f"    ❌ {friendly_name} failed after retries: {str(e)[:100]}")
            # Continue to next model
            continue

    # All models failed
    raise Exception("All models failed to process this page")


# =============================================================================
# OCR FUNCTION: G4F with Multiple Models & Mermaid Support
# =============================================================================
def extract_with_g4f_multimodel(pdf_path, output_dir='extracted_pages', models=OCR_MODELS):
    """
    Extracts text using G4F with multiple model fallback.
    Includes retry mechanism and Mermaid diagram generation.
    """
    os.makedirs(output_dir, exist_ok=True)

    # Initialize G4F Client
    client = Client()

    # Convert PDF to images
    print(f"🔄 Converting PDF to images...")
    try:
        pages = convert_from_path(pdf_path, dpi=300)
    except Exception as e:
        print(f"❌ Error converting PDF: {e}")
        return ""

    full_text = []
    model_stats = {model["name"]: 0 for model in models}

    # Enhanced OCR Prompt with Mermaid Support
    prompt = """You are a precision OCR system. Extract ONLY the actual text and content that appears in the image.

CRITICAL RULES:

1. OUTPUT FORMAT - EXTREMELY IMPORTANT:
   - Extract ONLY what you see in the image
   - DO NOT add page numbers unless they are IN the actual content
   - DO NOT add headers like "Page X" or "Extracted by:"
   - DO NOT add ANY metadata, footers, or commentary
   - DO NOT say "Here is the extracted text" or "The image contains"
   - DO NOT mention the OCR process at all
   - Start immediately with the first word/character from the image
   - If the page is blank, output nothing

2. MATHEMATICAL FORMATTING:
   - Inline math: $...$
   - Display math: $$...$$
   - Use strict LaTeX syntax (e.g., \\frac{}{}, \\int, \\sum, \\sqrt{}, \\alpha)
   - Preserve equation numbers if present

3. DIAGRAMS & VISUAL ELEMENTS (IMPORTANT):
   - For ANY diagrams, flowcharts, graphs, trees, or visual representations, create Mermaid code
   - Place Mermaid code in ```mermaid code blocks
   - Supported diagram types:
     * flowchart TD/LR (flowcharts, decision trees)
     * graph TD/LR (general graphs)
     * sequenceDiagram (sequence diagrams)
     * classDiagram (class diagrams, UML)
     * stateDiagram-v2 (state machines)
     * erDiagram (entity relationship)
     * pie (pie charts)
     * gantt (timelines, schedules)

   - Example flowchart:
     ```mermaid
     flowchart TD
         A[Start] --> B{Decision?}
         B -->|Yes| C[Process 1]
         B -->|No| D[Process 2]
         C --> E[End]
         D --> E
     ```

   - Example graph:
     ```mermaid
     graph LR
         A[Input] --> B[Processing]
         B --> C[Output]
     ```

   - For mathematical graphs/plots, describe axes, points, and relationships
   - If diagram is too complex for Mermaid, provide detailed textual description

4. STRUCTURE PRESERVATION:
   - Headers: Use Markdown (# ## ###)
   - Lists: Preserve numbered/bulleted lists exactly
   - Tables: Use Markdown table syntax
   - Bold/Italic: Use **bold** and *italic*
   - Code blocks: Use ```language for code snippets
   - Problem numbers: Maintain exact numbering (1.1, 1.2, etc.)

5. SPECIAL CONTENT:
   - Preserve page numbers, headers, footers
   - Note handwritten annotations: [Handwritten: text]
   - Indicate unclear text: [UNCLEAR: best guess]
   - Preserve figure captions and labels

Extract the content now."""

    for i, page in enumerate(pages):
        print(f"\n📄 Processing page {i+1}/{len(pages)}...")
        temp_img_path = f"temp_page_processing_{i}.png"

        try:
            # Save page to disk
            page.save(temp_img_path, format='PNG')

            # Try multiple models with fallback
            extracted_text, successful_model = extract_page_with_models(
                temp_img_path,
                client,
                models,
                prompt
            )

            # Clean any unwanted headers/metadata
            extracted_text = clean_extracted_text(extracted_text)

            # Track which model succeeded
            model_stats[successful_model] += 1

            full_text.append(extracted_text)
            print(f"  ✓ Page {i+1} completed ({len(extracted_text)} chars)")

            # Save Markdown immediately (ONLY content, no headers)
            page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
            with open(page_file, 'w', encoding='utf-8') as f:
                f.write(extracted_text)
            print(f"  💾 Saved: {page_file} (Model: {successful_model})")

            # Rate limiting delay between pages
            if i < len(pages) - 1:
                time.sleep(PAGE_DELAY)

        except Exception as e:
            error_msg = f"[ERROR on page {i+1}: All {len(models)} models failed after {MAX_RETRIES} attempts each]\n\n{str(e)}"
            print(f"  ✗ {error_msg[:200]}")
            full_text.append(error_msg)

            # Save error log
            page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
            with open(page_file, 'w', encoding='utf-8') as f:
                f.write(f"# Page {i+1}\n\n")
                f.write(error_msg)

        finally:
            # Cleanup temp file
            if os.path.exists(temp_img_path):
                os.remove(temp_img_path)

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text), model_stats


# =============================================================================
# BATCH PROCESSING
# =============================================================================
def extract_large_pdf_batch(pdf_path, output_file='extracted_text.md', output_dir='extracted_pages', models=OCR_MODELS):
    """
    Process PDF with memory management, multi-model fallback, and status updates
    """
    print("="*70)
    print("📚 PDF OCR EXTRACTION WITH MULTI-MODEL FALLBACK & MERMAID SUPPORT")
    print("="*70)
    print(f"📄 PDF: {pdf_path}")
    print(f"📁 Output directory: {output_dir}")
    print(f"🔄 Max retries per model: {MAX_RETRIES}")
    print(f"⏱️  Retry delay: {RETRY_DELAY}s (exponential backoff)")
    print(f"🤖 Available models: {len(models)}")
    for idx, model in enumerate(models, 1):
        print(f"   {idx}. {model['name']} ({model['model']})")
    print("="*70 + "\n")

    start_time = time.time()

    # Execute extraction
    text, model_stats = extract_with_g4f_multimodel(pdf_path, output_dir, models)

    # Save combined text (content only, metadata in comments)
    with open(output_file, 'w', encoding='utf-8') as f:
        # Optional: Add metadata as HTML comments (invisible in rendered Markdown)
        f.write(f"<!-- Extracted from: {pdf_path} -->\n")
        f.write(f"<!-- Extraction date: {time.strftime('%Y-%m-%d %H:%M:%S')} -->\n")
        f.write(f"<!-- Models used: {', '.join([m for m, c in model_stats.items() if c > 0])} -->\n\n")

        # Write actual content with page separators
        f.write(text)

    # Count pages
    if os.path.exists(output_dir):
        page_files = sorted([f for f in os.listdir(output_dir) if f.startswith('page_') and f.endswith('.md')])
        successful_pages = len([f for f in page_files if 'ERROR' not in open(os.path.join(output_dir, f)).read()])
    else:
        page_files = []
        successful_pages = 0

    elapsed_time = time.time() - start_time

    print("\n" + "="*70)
    print("✅ EXTRACTION COMPLETE!")
    print("="*70)
    print(f"📊 Total pages: {len(page_files)}")
    print(f"✅ Successfully extracted: {successful_pages}")
    print(f"❌ Failed pages: {len(page_files) - successful_pages}")
    print(f"📝 Combined file: {output_file}")
    print(f"⏱️  Time elapsed: {elapsed_time:.1f}s ({elapsed_time/len(page_files):.1f}s per page)")
    print("\n🤖 Model Usage:")
    for model_name, count in sorted(model_stats.items(), key=lambda x: x[1], reverse=True):
        if count > 0:
            percentage = (count / len(page_files)) * 100
            print(f"   • {model_name}: {count} pages ({percentage:.1f}%)")
    print("="*70 + "\n")

    # Clean up memory
    gc.collect()

    return text


# =============================================================================
# FILE MANAGEMENT UTILITIES
# =============================================================================
def list_pdf_files(directory='.'):
    """List all PDF files in the specified directory"""
    pdf_files = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.lower().endswith('.pdf'):
                pdf_files.append(os.path.join(root, file))
    return pdf_files


def mount_google_drive():
    """Mount Google Drive if not already mounted"""
    try:
        from google.colab import drive
        if not os.path.exists('/content/drive'):
            print("📂 Mounting Google Drive...")
            drive.mount('/content/drive')
            print("✅ Google Drive mounted successfully!")
            return True
        else:
            print("✅ Google Drive already mounted")
            return True
    except Exception as e:
        print(f"❌ Could not mount Google Drive: {e}")
        return False


def select_pdf_file():
    """Interactive file selection with multiple options"""
    print("="*70)
    print("🚀 MULTI-MODEL PDF OCR EXTRACTOR")
    print("="*70)
    print("\n📋 This tool will try multiple AI models in sequence:")
    print("   1️⃣  DeepSeek-OCR (specialized OCR)")
    print("   2️⃣  OLM-OCR (specialized OCR)")
    print("   3️⃣  PaddleOCR (specialized OCR)")
    print("   4️⃣  Phi-4-Multimodal (general vision)")
    print("   5️⃣  Opera-Aria (general vision)")
    print("\n💡 Each model gets 3 retry attempts before moving to the next")
    print("📊 Supports LaTeX math and Mermaid diagrams")
    print("="*70 + "\n")

    print("📁 FILE SELECTION OPTIONS:")
    print("="*70)
    print("1️⃣  Use existing file in Colab (enter path)")
    print("2️⃣  List PDF files in current directory")
    print("3️⃣  Mount Google Drive and select file")
    print("4️⃣  Upload new file")
    print("="*70)

    choice = input("\n👉 Enter your choice (1-4): ").strip()

    if choice == '1':
        # Direct path input
        file_path = input("📝 Enter PDF file path: ").strip()
        if os.path.exists(file_path):
            print(f"✅ Found: {file_path}")
            return file_path
        else:
            print(f"❌ File not found: {file_path}")
            return None

    elif choice == '2':
        # List files in current directory
        print("\n🔍 Scanning for PDF files...")
        pdf_files = list_pdf_files('.')

        if not pdf_files:
            print("❌ No PDF files found in current directory")
            return None

        print(f"\n📚 Found {len(pdf_files)} PDF file(s):")
        print("="*70)
        for idx, pdf in enumerate(pdf_files, 1):
            size_mb = os.path.getsize(pdf) / (1024 * 1024)
            print(f"{idx}. {pdf} ({size_mb:.2f} MB)")
        print("="*70)

        selection = input(f"\n👉 Select file number (1-{len(pdf_files)}): ").strip()
        try:
            file_idx = int(selection) - 1
            if 0 <= file_idx < len(pdf_files):
                selected_file = pdf_files[file_idx]
                print(f"✅ Selected: {selected_file}")
                return selected_file
            else:
                print("❌ Invalid selection")
                return None
        except ValueError:
            print("❌ Invalid input")
            return None

    elif choice == '3':
        # Google Drive
        if mount_google_drive():
            print("\n🔍 Scanning Google Drive for PDF files...")
            pdf_files = list_pdf_files('/content/drive/MyDrive')

            if not pdf_files:
                print("❌ No PDF files found in Google Drive")
                return None

            # Show only first 20 files to avoid clutter
            display_files = pdf_files[:20]
            print(f"\n📚 Found {len(pdf_files)} PDF file(s) (showing first {len(display_files)}):")
            print("="*70)
            for idx, pdf in enumerate(display_files, 1):
                size_mb = os.path.getsize(pdf) / (1024 * 1024)
                # Show relative path from MyDrive
                rel_path = pdf.replace('/content/drive/MyDrive/', '')
                print(f"{idx}. {rel_path} ({size_mb:.2f} MB)")
            print("="*70)

            if len(pdf_files) > 20:
                print(f"\n💡 {len(pdf_files) - 20} more files available")
                print("💡 Or enter full path manually")

            selection = input(f"\n👉 Select file number (1-{len(display_files)}) or enter path: ").strip()

            # Try as number first
            try:
                file_idx = int(selection) - 1
                if 0 <= file_idx < len(display_files):
                    selected_file = display_files[file_idx]
                    print(f"✅ Selected: {selected_file}")
                    return selected_file
            except ValueError:
                # Try as path
                if selection.startswith('/'):
                    file_path = selection
                else:
                    file_path = f"/content/drive/MyDrive/{selection}"

                if os.path.exists(file_path):
                    print(f"✅ Found: {file_path}")
                    return file_path
                else:
                    print(f"❌ File not found: {file_path}")
                    return None
        return None

    elif choice == '4':
        # Upload file
        print("\n📤 Upload your PDF file:")
        uploaded = files.upload()
        if uploaded:
            pdf_filename = list(uploaded.keys())[0]
            print(f"✅ Uploaded: {pdf_filename}")
            return pdf_filename
        else:
            print("❌ No file uploaded")
            return None

    else:
        print("❌ Invalid choice")
        return None


# =============================================================================
# MAIN EXECUTION
# =============================================================================
if __name__ == "__main__":
    # Interactive file selection
    pdf_filename = select_pdf_file()

    if pdf_filename:
        # Get output filename based on input
        base_name = os.path.splitext(os.path.basename(pdf_filename))[0]
        output_file = f"{base_name}_extracted.md"
        output_dir = f"{base_name}_pages"

        print(f"\n📂 Output will be saved as:")
        print(f"   • Combined file: {output_file}")
        print(f"   • Individual pages: {output_dir}/")

        confirm = input("\n👉 Proceed with extraction? (y/n): ").strip().lower()

        if confirm == 'y':
            # Run Extraction with multi-model fallback
            text = extract_large_pdf_batch(
                pdf_path=pdf_filename,
                output_file=output_file,
                output_dir=output_dir,
                models=OCR_MODELS
            )

            # Preview
            print("\n" + "="*70)
            print("👀 PREVIEW (first 1000 characters):")
            print("="*70)
            print(text[:1000])
            print("\n[...truncated...]")

            # Download
            print("\n📥 Downloading combined file...")
            files.download(output_file)

            print("\n💡 TIPS:")
            print(f"   • Individual pages saved in '{output_dir}/' folder")
            print("   • Each page shows which model extracted it")
            print("   • Mermaid diagrams can be rendered in Markdown viewers")
            print("   • LaTeX math can be rendered in Markdown viewers")
            print(f"   • File is also saved in Colab at: {output_file}")
        else:
            print("❌ Extraction cancelled")
    else:
        print("\n❌ No valid file selected. Please run the script again.")
````

````python
"""
Zip the extracted_pages folder after PDF OCR extraction
"""

import zipfile
import os
from pathlib import Path
from datetime import datetime


def zip_extracted_pages(
    folder_path='extracted_pages',
    output_zip=None,
    include_combined_file=True,
    combined_file='calculus_extracted.md'
):
    """
    Zip the extracted pages folder

    Args:
        folder_path: Path to the folder containing extracted pages
        output_zip: Name of output zip file (auto-generated if None)
        include_combined_file: Whether to include the combined markdown file
        combined_file: Path to the combined markdown file

    Returns:
        Path to the created zip file
    """

    # Check if folder exists
    if not os.path.exists(folder_path):
        raise FileNotFoundError(f"Folder not found: {folder_path}")

    # Auto-generate zip filename if not provided
    if output_zip is None:
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        output_zip = f"extracted_pages_{timestamp}.zip"

    # Ensure .zip extension
    if not output_zip.endswith('.zip'):
        output_zip += '.zip'

    print(f"Creating zip file: {output_zip}")
    print(f"Source folder: {folder_path}")

    file_count = 0
    total_size = 0

    with zipfile.ZipFile(output_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
        # Add all files from the folder
        for root, dirs, files in os.walk(folder_path):
            for file in sorted(files):
                file_path = os.path.join(root, file)
                # Create relative path for the zip archive
                arcname = os.path.relpath(file_path, os.path.dirname(folder_path))

                zipf.write(file_path, arcname)
                file_size = os.path.getsize(file_path)
                total_size += file_size
                file_count += 1
                print(f"  ✓ Added: {arcname} ({file_size:,} bytes)")

        # Add combined file if it exists and requested
        if include_combined_file and os.path.exists(combined_file):
            zipf.write(combined_file, os.path.basename(combined_file))
            file_size = os.path.getsize(combined_file)
            total_size += file_size
            file_count += 1
            print(f"  ✓ Added: {os.path.basename(combined_file)} ({file_size:,} bytes)")

    zip_size = os.path.getsize(output_zip)
    compression_ratio = (1 - zip_size / total_size) * 100 if total_size > 0 else 0

    print(f"\n✓ Zip created successfully!")
    print(f"✓ Files added: {file_count}")
    print(f"✓ Original size: {total_size:,} bytes")
    print(f"✓ Compressed size: {zip_size:,} bytes")
    print(f"✓ Compression ratio: {compression_ratio:.1f}%")
    print(f"✓ Output: {output_zip}")

    return output_zip


def zip_with_custom_name(folder_path='extracted_pages', pdf_name='document'):
    """
    Create a zip file named after the PDF document

    Args:
        folder_path: Path to the folder containing extracted pages
        pdf_name: Name of the original PDF (without extension)

    Returns:
        Path to the created zip file
    """
    output_zip = f"{pdf_name}_extracted.zip"
    return zip_extracted_pages(folder_path, output_zip)


def zip_multiple_folders(folders, output_zip='all_extracted_pages.zip'):
    """
    Zip multiple extraction folders into one archive

    Args:
        folders: List of folder paths to include
        output_zip: Name of the output zip file

    Returns:
        Path to the created zip file
    """
    print(f"Creating zip file: {output_zip}")

    file_count = 0
    total_size = 0

    with zipfile.ZipFile(output_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for folder in folders:
            if not os.path.exists(folder):
                print(f"  ⚠ Skipping non-existent folder: {folder}")
                continue

            print(f"\nAdding folder: {folder}")

            for root, dirs, files in os.walk(folder):
                for file in sorted(files):
                    file_path = os.path.join(root, file)
                    # Keep folder structure in zip
                    arcname = os.path.relpath(file_path, os.path.dirname(folders[0]))

                    zipf.write(file_path, arcname)
                    file_size = os.path.getsize(file_path)
                    total_size += file_size
                    file_count += 1
                    print(f"  ✓ Added: {arcname}")

    zip_size = os.path.getsize(output_zip)

    print(f"\n✓ Zip created successfully!")
    print(f"✓ Files added: {file_count}")
    print(f"✓ Compressed size: {zip_size:,} bytes")
    print(f"✓ Output: {output_zip}")

    return output_zip


# =============================================================================
# GOOGLE COLAB SPECIFIC
# =============================================================================
def zip_and_download_colab(folder_path='extracted_pages', output_zip=None):
    """
    Create zip file and automatically download in Google Colab

    Args:
        folder_path: Path to the folder containing extracted pages
        output_zip: Name of output zip file (auto-generated if None)
    """
    try:
        from google.colab import files

        # Create the zip file
        zip_file = zip_extracted_pages(folder_path, output_zip)

        # Download it
        print(f"\nDownloading {zip_file}...")
        files.download(zip_file)
        print("✓ Download started!")

        return zip_file

    except ImportError:
        print("⚠ Not running in Google Colab. Zip file created but not auto-downloaded.")
        return zip_extracted_pages(folder_path, output_zip)


# =============================================================================
# MAIN EXECUTION
# =============================================================================
if __name__ == "__main__":
    import sys

    # Check if running in Colab
    try:
        from google.colab import files
        IN_COLAB = True
    except ImportError:
        IN_COLAB = False

    # Example usage based on environment
    if IN_COLAB:
        # In Google Colab - zip and auto-download
        print("Running in Google Colab")
        print("="*50)

        # Simple usage - zip extracted_pages folder
        zip_and_download_colab('extracted_pages')

    else:
        # Local execution - just create the zip
        print("Running locally")
        print("="*50)

        # Check command line arguments
        if len(sys.argv) > 1:
            folder = sys.argv[1]
            output = sys.argv[2] if len(sys.argv) > 2 else None
            zip_extracted_pages(folder, output)
        else:
            # Default - zip extracted_pages folder
            zip_extracted_pages('extracted_pages')

        print("\n" + "="*50)
        print("USAGE EXAMPLES:")
        print("="*50)
        print("1. Zip default folder:")
        print("   python zip_extracted_pages.py")
        print("\n2. Zip specific folder:")
        print("   python zip_extracted_pages.py my_pages")
        print("\n3. Zip with custom output name:")
        print("   python zip_extracted_pages.py extracted_pages my_output.zip")
        print("\n4. Use in Python script:")
        print("   from zip_extracted_pages import zip_extracted_pages")
        print("   zip_extracted_pages('extracted_pages', 'output.zip')")
````
