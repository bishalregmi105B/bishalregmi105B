```python
"""
Advanced PDF OCR for Mathematical Content - Google Colab
Multiple methods for high-accuracy text extraction from math PDFs
"""

# =============================================================================
# METHOD 1: Google Cloud Vision API (BEST for complex math)
# =============================================================================
def extract_with_google_vision(pdf_path, output_dir='extracted_pages'):
    """
    Most accurate for complex mathematical content
    Requires: Google Cloud account with Vision API enabled
    """
    from google.cloud import vision
    from pdf2image import convert_from_path
    import io
    import os

    os.makedirs(output_dir, exist_ok=True)
    client = vision.ImageAnnotatorClient()
    pages = convert_from_path(pdf_path, dpi=300)

    full_text = []

    for i, page in enumerate(pages):
        print(f"Processing page {i+1}/{len(pages)}...")

        # Convert PIL image to bytes
        img_byte_arr = io.BytesIO()
        page.save(img_byte_arr, format='PNG')
        img_byte_arr = img_byte_arr.getvalue()

        image = vision.Image(content=img_byte_arr)
        response = client.document_text_detection(image=image)

        if response.error.message:
            raise Exception(f'{response.error.message}')

        text = response.full_text_annotation.text
        full_text.append(text)

        # Save page immediately
        page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
        with open(page_file, 'w', encoding='utf-8') as f:
            f.write(text)
        print(f"  ✓ Saved: {page_file}")

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text)


# =============================================================================
# METHOD 2: PaddleOCR (FREE, excellent for math)
# =============================================================================
def extract_with_paddleocr(pdf_path, output_dir='extracted_pages'):
    """
    Free and powerful OCR with good math support
    Install: !pip install paddlepaddle paddleocr
    """
    from paddleocr import PaddleOCR
    from pdf2image import convert_from_path
    import numpy as np
    import os

    os.makedirs(output_dir, exist_ok=True)
    # Initialize with English language
    ocr = PaddleOCR(use_angle_cls=True, lang='en', use_gpu=True)

    pages = convert_from_path(pdf_path, dpi=300)
    full_text = []

    for i, page in enumerate(pages):
        print(f"Processing page {i+1}/{len(pages)}...")

        # Convert to numpy array
        img_array = np.array(page)

        # Perform OCR
        result = ocr.ocr(img_array, cls=True)

        # Extract text from result
        page_text = []
        if result and result[0]:
            for line in result[0]:
                if line[1][0]:  # Check if text exists
                    page_text.append(line[1][0])

        text = '\n'.join(page_text)
        full_text.append(text)

        # Save page immediately
        page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
        with open(page_file, 'w', encoding='utf-8') as f:
            f.write(text)
        print(f"  ✓ Saved: {page_file}")

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text)


# =============================================================================
# METHOD 3: EasyOCR (FREE, good balance)
# =============================================================================
def extract_with_easyocr(pdf_path, output_dir='extracted_pages'):
    """
    Free, easy to use, decent accuracy
    Install: !pip install easyocr
    """
    import easyocr
    from pdf2image import convert_from_path
    import numpy as np
    import os

    os.makedirs(output_dir, exist_ok=True)
    reader = easyocr.Reader(['en'], gpu=True)
    pages = convert_from_path(pdf_path, dpi=300)

    full_text = []

    for i, page in enumerate(pages):
        print(f"Processing page {i+1}/{len(pages)}...")

        img_array = np.array(page)
        result = reader.readtext(img_array, detail=0, paragraph=True)

        text = '\n'.join(result)
        full_text.append(text)

        # Save page immediately
        page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
        with open(page_file, 'w', encoding='utf-8') as f:
            f.write(text)
        print(f"  ✓ Saved: {page_file}")

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text)


# =============================================================================
# METHOD 4: Tesseract with Advanced Config (IMPROVED)
# =============================================================================
def extract_with_advanced_tesseract(pdf_path, output_dir='extracted_pages'):
    """
    Enhanced Tesseract with optimal settings for math
    Note: Basic OCR - does not format as Markdown. Use G4F methods for Markdown output.
    """
    import pytesseract
    from pdf2image import convert_from_path
    from PIL import ImageEnhance, ImageFilter
    import os

    os.makedirs(output_dir, exist_ok=True)
    # Custom config for better accuracy
    custom_config = r'--oem 3 --psm 6 -c preserve_interword_spaces=1'

    pages = convert_from_path(pdf_path, dpi=300)  # Higher DPI
    full_text = []

    for i, page in enumerate(pages):
        print(f"Processing page {i+1}/{len(pages)}...")

        # Image preprocessing for better OCR
        # Convert to grayscale
        page = page.convert('L')

        # Enhance contrast
        enhancer = ImageEnhance.Contrast(page)
        page = enhancer.enhance(2.0)

        # Sharpen
        page = page.filter(ImageFilter.SHARPEN)

        # Extract text
        text = pytesseract.image_to_string(page, config=custom_config)
        full_text.append(text)

        # Save page immediately
        page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
        with open(page_file, 'w', encoding='utf-8') as f:
            f.write(text)
        print(f"  ✓ Saved: {page_file}")

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text)


# =============================================================================
# METHOD 5: Mathpix API (SPECIALIZED for math LaTeX)
# =============================================================================
def extract_with_mathpix(pdf_path, output_dir='extracted_pages'):
    """
    Best for converting math equations to LaTeX
    Requires: Mathpix API key (paid but has free tier)
    Note: Returns LaTeX, you may need to wrap in $ or $ for KaTeX rendering
    """
    import requests
    from pdf2image import convert_from_path
    import io
    import base64
    import os

    os.makedirs(output_dir, exist_ok=True)
    # Get your API keys from https://mathpix.com/
    APP_ID = 'your_app_id'
    APP_KEY = 'your_app_key'

    pages = convert_from_path(pdf_path, dpi=300)
    full_text = []

    for i, page in enumerate(pages):
        print(f"Processing page {i+1}/{len(pages)}...")

        # Convert to bytes
        img_byte_arr = io.BytesIO()
        page.save(img_byte_arr, format='PNG')
        img_byte_arr = img_byte_arr.getvalue()

        # Encode to base64
        image_uri = "data:image/png;base64," + base64.b64encode(img_byte_arr).decode()

        # Make API request - requesting markdown format
        response = requests.post(
            "https://api.mathpix.com/v3/text",
            json={
                "src": image_uri,
                "formats": ["text", "latex_styled", "markdown"],  # Request markdown
                "math_inline_delimiters": ["$", "$"],
                "math_display_delimiters": ["$", "$"],
                "markdown": True
            },
            headers={
                "app_id": APP_ID,
                "app_key": APP_KEY
            }
        )

        if response.ok:
            result = response.json()
            # Prefer markdown format if available, fallback to text
            text = result.get('markdown', result.get('text', ''))
            full_text.append(text)

            # Save page immediately
            page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
            with open(page_file, 'w', encoding='utf-8') as f:
                f.write(text)
            print(f"  ✓ Saved: {page_file}")
        else:
            print(f"Error on page {i+1}: {response.status_code}")

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text)


# =============================================================================
# METHOD 6: Claude API with Vision (EXCELLENT for understanding context)
# =============================================================================
def extract_with_claude(pdf_path, output_dir='extracted_pages'):
    """
    Use Claude's vision capabilities to extract and understand math content
    Outputs properly formatted Markdown with KaTeX/LaTeX
    """
    import anthropic
    from pdf2image import convert_from_path
    import base64
    import io
    import os

    os.makedirs(output_dir, exist_ok=True)
    client = anthropic.Anthropic(api_key='your_api_key')
    pages = convert_from_path(pdf_path, dpi=300)

    full_text = []

    for i, page in enumerate(pages):
        print(f"Processing page {i+1}/{len(pages)}...")

        # Convert to base64
        img_byte_arr = io.BytesIO()
        page.save(img_byte_arr, format='PNG')
        img_byte_arr = img_byte_arr.getvalue()
        base64_image = base64.b64encode(img_byte_arr).decode()

        message = client.messages.create(
            model="claude-sonnet-4-20250514",
            max_tokens=4000,
            messages=[{
                "role": "user",
                "content": [
                    {
                        "type": "image",
                        "source": {
                            "type": "base64",
                            "media_type": "image/png",
                            "data": base64_image
                        }
                    },
                    {
                        "type": "text",
                        "text": """Extract all text from this page as MARKDOWN with KaTeX/LaTeX:

FORMAT REQUIREMENTS:
1. **Markdown structure**: Use # headers, **bold**, numbered lists, bullet points
2. **Math notation**:
   - Inline math: $x^2 + 1$
   - Display/block math: $\\int_0^1 f(x)dx$
   - Use proper LaTeX: \\frac{a}{b}, \\sqrt{x}, \\sin, \\cos, \\sum, \\int, \\lim
3. **Organization**: Clear problem numbers, show all steps, include explanations
4. **Precision**: Exact subscripts, superscripts, operators, all mathematical symbols

Be EXTREMELY accurate with mathematical notation. Output ONLY the formatted markdown."""
                    }
                ]
            }]
        )

        text = message.content[0].text
        full_text.append(text)

        # Save page immediately
        page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
        with open(page_file, 'w', encoding='utf-8') as f:
            f.write(text)
        print(f"  ✓ Saved: {page_file}")

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text)


# =============================================================================
# METHOD 7: G4F (GPT4Free) - FREE Vision Models (EXCELLENT & FREE!)
# =============================================================================
def extract_with_g4f(pdf_path, provider_name='Blackbox', output_dir='extracted_pages'):
    """
    Use G4F (GPT4Free) vision models - COMPLETELY FREE!
    Install: !pip install g4f

    Available providers with vision support:
    - Blackbox (Recommended - Fast & Reliable)
    - Copilot (Microsoft - Very Good)
    - Gemini (Google - Excellent for complex tasks)
    - DeepSeek (Good accuracy)
    - HuggingChat (Open source models)
    """
    import g4f
    from g4f.client import Client
    from pdf2image import convert_from_path
    import io
    import os

    os.makedirs(output_dir, exist_ok=True)
    pages = convert_from_path(pdf_path, dpi=300)
    full_text = []

    # Initialize G4F client with provider
    provider_map = {
        'Blackbox': g4f.Provider.Blackbox,
        'Copilot': g4f.Provider.Copilot,
        'Gemini': g4f.Provider.Gemini,
        'DeepSeek': g4f.Provider.DeepSeek,
        'HuggingChat': g4f.Provider.HuggingChat
    }

    provider = provider_map.get(provider_name, g4f.Provider.Blackbox)
    client = Client(provider=provider)

    for i, page in enumerate(pages):
        print(f"Processing page {i+1}/{len(pages)} using {provider_name}...")

        try:
            # Convert PIL image to bytes (file-like object)
            img_byte_arr = io.BytesIO()
            page.save(img_byte_arr, format='PNG')
            img_byte_arr.seek(0)  # Reset pointer to beginning

            # Create prompt for mathematical content extraction
            prompt = """You are a precise OCR system. Extract all text from this image exactly as it appears.

CRITICAL RULES:
- Output ONLY the actual text from the image
- DO NOT add any commentary, explanations, or meta-text
- DO NOT write phrases like "Here is the extracted text:", "The image contains:", etc.
- DO NOT add introductions or conclusions
- Start immediately with the actual content

FORMATTING:
1. Mathematical equations: Use LaTeX format ($equation$ for inline, $$equation$$ for display)
2. Preserve exact structure, problem numbers, and formatting
3. Be extremely precise with subscripts, superscripts, fractions, integrals, limits
4. Include all explanatory text and annotations exactly as shown

Output the raw content only."""

            # G4F API call with image
            response = client.chat.completions.create(
                model="",  # Provider's default model
                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],
                image=img_byte_arr  # Pass file-like object directly
            )

            # Extract response text
            extracted_text = response.choices[0].message.content
            full_text.append(extracted_text)
            print(f"  ✓ Page {i+1} completed ({len(extracted_text)} chars)")

            # Save page immediately
            page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
            with open(page_file, 'w', encoding='utf-8') as f:
                f.write(extracted_text)
            print(f"  ✓ Saved: {page_file}")

        except Exception as e:
            print(f"  ✗ Error on page {i+1}: {str(e)}")
            # Try fallback to Copilot
            try:
                print(f"  Retrying with Copilot...")
                img_byte_arr.seek(0)
                client_fallback = Client(provider=g4f.Provider.Copilot)
                response = client_fallback.chat.completions.create(
                    model="",
                    messages=[{"role": "user", "content": prompt}],
                    image=img_byte_arr
                )
                extracted_text = response.choices[0].message.content
                full_text.append(extracted_text)
                print(f"  ✓ Fallback succeeded")

                # Save page from fallback
                page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
                with open(page_file, 'w', encoding='utf-8') as f:
                    f.write(extracted_text)
                print(f"  ✓ Saved: {page_file}")
            except Exception as e2:
                print(f"  ✗ Fallback also failed: {str(e2)}")
                error_text = f"[ERROR: Could not process page {i+1}]"
                full_text.append(error_text)
                # Save error page
                page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
                with open(page_file, 'w', encoding='utf-8') as f:
                    f.write(error_text)

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text)


# =============================================================================
# METHOD 8: G4F Multi-Provider Ensemble (BEST ACCURACY - FREE!)
# =============================================================================
def extract_with_g4f_ensemble(pdf_path, providers=['Blackbox', 'Copilot', 'Gemini'], output_dir='extracted_pages'):
    """
    Use multiple G4F providers and combine results for maximum accuracy
    This method queries multiple AI models and picks the longest/most detailed result
    """
    import g4f
    from g4f.client import Client
    from pdf2image import convert_from_path
    import io
    import os

    os.makedirs(output_dir, exist_ok=True)
    pages = convert_from_path(pdf_path, dpi=300)
    full_text = []

    # Provider configurations
    provider_map = {
        'Blackbox': g4f.Provider.Blackbox,
        'Copilot': g4f.Provider.Copilot,
        'Gemini': g4f.Provider.Gemini,
        'DeepSeek': g4f.Provider.DeepSeek,
        'HuggingChat': g4f.Provider.HuggingChat
    }

    prompt = """You are a precise OCR system. Extract all text from this image exactly as it appears.

CRITICAL RULES:
- Output ONLY the actual text from the image
- DO NOT add any commentary, explanations, or meta-text
- DO NOT write phrases like "Here is the extracted text:", "The image contains:", etc.
- DO NOT add introductions or conclusions
- Start immediately with the actual content

FORMATTING:
- Mathematical equations in LaTeX format ($inline$ or $$display$$)
- All numbers, variables, operators, subscripts, superscripts
- All problem statements, solutions, and explanations
- Preserve exact formatting and structure

Output the raw content only."""

    for i, page in enumerate(pages):
        print(f"\nProcessing page {i+1}/{len(pages)} with {len(providers)} providers...")

        # Convert page to bytes once
        img_byte_arr = io.BytesIO()
        page.save(img_byte_arr, format='PNG')

        # Query multiple models
        results = []
        for provider_name in providers:
            if provider_name not in provider_map:
                continue

            provider = provider_map[provider_name]
            try:
                print(f"  Querying {provider_name}...", end=" ")

                # Reset buffer position for each provider
                img_byte_arr.seek(0)

                # Create client and query
                client = Client(provider=provider)
                response = client.chat.completions.create(
                    model="",
                    messages=[{"role": "user", "content": prompt}],
                    image=img_byte_arr
                )

                extracted = response.choices[0].message.content
                results.append({
                    'provider': provider_name,
                    'text': extracted,
                    'length': len(extracted)
                })
                print(f"✓ ({len(extracted)} chars)")

            except Exception as e:
                print(f"✗ Failed: {str(e)[:50]}")

        # Combine results (use longest response as it likely has most detail)
        if results:
            # Sort by length and get the longest
            best_result = max(results, key=lambda x: x['length'])
            print(f"  → Best result: {best_result['provider']} ({best_result['length']} chars)")
            full_text.append(best_result['text'])

            # Save page immediately
            page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
            with open(page_file, 'w', encoding='utf-8') as f:
                f.write(best_result['text'])
            print(f"  ✓ Saved: {page_file}")
        else:
            print(f"  ✗ All providers failed for page {i+1}")
            error_text = f"[ERROR: All providers failed for page {i+1}]"
            full_text.append(error_text)
            # Save error page
            page_file = os.path.join(output_dir, f'page_{i+1:03d}.md')
            with open(page_file, 'w', encoding='utf-8') as f:
                f.write(error_text)

    return '\n\n--- PAGE BREAK ---\n\n'.join(full_text)


# =============================================================================
# BATCH PROCESSING with Progress Tracking
# =============================================================================
def extract_large_pdf_batch(pdf_path, method='paddleocr', output_file='extracted_text.txt', output_dir='extracted_pages'):
    """
    Process large PDFs with progress tracking and memory management
    Saves each page as a separate markdown file concurrently
    """
    from tqdm import tqdm
    import gc
    import os

    print(f"Starting extraction using {method}...")
    print(f"PDF: {pdf_path}")
    print(f"Output directory: {output_dir}")
    print(f"Combined output file: {output_file}")

    # Choose method
    if method == 'google_vision':
        text = extract_with_google_vision(pdf_path, output_dir)
    elif method == 'paddleocr':
        text = extract_with_paddleocr(pdf_path, output_dir)
    elif method == 'easyocr':
        text = extract_with_easyocr(pdf_path, output_dir)
    elif method == 'tesseract':
        text = extract_with_advanced_tesseract(pdf_path, output_dir)
    elif method == 'mathpix':
        text = extract_with_mathpix(pdf_path, output_dir)
    elif method == 'claude':
        text = extract_with_claude(pdf_path, output_dir)
    elif method == 'g4f':
        text = extract_with_g4f(pdf_path, provider_name='Blackbox', output_dir=output_dir)
    elif method == 'g4f_ensemble':
        text = extract_with_g4f_ensemble(pdf_path, providers=['Blackbox', 'Copilot', 'Gemini'], output_dir=output_dir)
    else:
        raise ValueError(f"Unknown method: {method}")

    # Save combined text to file
    with open(output_file, 'w', encoding='utf-8') as f:
        f.write(text)

    # Count saved pages
    page_files = sorted([f for f in os.listdir(output_dir) if f.startswith('page_') and f.endswith('.md')])

    print(f"\n✓ Extraction complete!")
    print(f"✓ Individual pages saved to: {output_dir}/ ({len(page_files)} pages)")
    print(f"✓ Combined file saved to: {output_file}")
    print(f"✓ Total characters: {len(text)}")

    # Clean up memory
    gc.collect()

    return text


# =============================================================================
# MAIN EXECUTION - GOOGLE COLAB
# =============================================================================
if __name__ == "__main__":
    # Install required packages (run in Colab)
    """
    # Basic OCR tools
    !apt-get install -y poppler-utils tesseract-ocr
    !pip install pdf2image pytesseract pillow tqdm

    # Advanced OCR (choose what you need)
    !pip install paddlepaddle paddleocr  # PaddleOCR - Recommended!
    !pip install easyocr  # EasyOCR

    # AI Vision Models - FREE with G4F! (HIGHLY RECOMMENDED)
    !pip install g4f  # G4F - Access multiple AI vision models FREE!

    # For Google Vision (if using - requires Google Cloud account)
    !pip install google-cloud-vision

    # For Anthropic Claude (if using - requires API key)
    !pip install anthropic

    # For Mathpix (if using - requires API key, has free tier)
    # Already has requests in Colab
    """

    # Upload your PDF in Colab
    from google.colab import files
    print("Upload your PDF file:")
    uploaded = files.upload()
    pdf_filename = list(uploaded.keys())[0]

    # =========================================================================
    # CHOOSE YOUR METHOD:
    # =========================================================================

    # OPTION 1: G4F Ensemble (BEST - Multiple AI models, FREE, Markdown + KaTeX output!)
    # Uses Gemini + Copilot + Blackbox together for maximum accuracy
    # text = extract_large_pdf_batch(
    #     pdf_path=pdf_filename,
    #     method='g4f_ensemble',
    #     output_file='calculus_extracted.md',  # .md extension for markdown
    #     output_dir='extracted_pages'  # Each page saved separately here
    # )

    # OPTION 2: G4F Single Model (FREE, very good, Markdown + KaTeX)
    text = extract_large_pdf_batch(
        pdf_path=pdf_filename,
        method='g4f',  # Uses Blackbox by default
        output_file='calculus_extracted.md',
        output_dir='extracted_pages'  # Each page saved separately here
    )

    # OPTION 3: Claude API (Excellent accuracy, Markdown + KaTeX)
    # text = extract_large_pdf_batch(
    #     pdf_path=pdf_filename,
    #     method='claude',  # Requires API key
    #     output_file='calculus_extracted.md',
    #     output_dir='extracted_pages'
    # )

    # OPTION 4: Mathpix (Best for LaTeX, has markdown support)
    # text = extract_large_pdf_batch(
    #     pdf_path=pdf_filename,
    #     method='mathpix',  # Requires API key
    #     output_file='calculus_extracted.md',
    #     output_dir='extracted_pages'
    # )

    # OPTION 5: PaddleOCR (FREE, traditional OCR, fast - NO markdown formatting)
    # text = extract_large_pdf_batch(
    #     pdf_path=pdf_filename,
    #     method='paddleocr',
    #     output_file='calculus_extracted.txt',
    #     output_dir='extracted_pages'
    # )

    # Other methods (not recommended for web rendering):
    # method='google_vision'  - Most accurate but NO markdown formatting
    # method='easyocr'        - Good balance but NO markdown formatting
    # method='tesseract'      - Basic OCR, NO markdown formatting

    # Display first 1000 characters
    print("\n" + "="*50)
    print("PREVIEW OF EXTRACTED MARKDOWN:")
    print("="*50)
    print(text[:1000])
    print("...")

    # Download the result
    files.download('calculus_extracted.md')
```

```python
"""
Zip the extracted_pages folder after PDF OCR extraction
"""

import zipfile
import os
from pathlib import Path
from datetime import datetime


def zip_extracted_pages(
    folder_path='extracted_pages',
    output_zip=None,
    include_combined_file=True,
    combined_file='calculus_extracted.md'
):
    """
    Zip the extracted pages folder

    Args:
        folder_path: Path to the folder containing extracted pages
        output_zip: Name of output zip file (auto-generated if None)
        include_combined_file: Whether to include the combined markdown file
        combined_file: Path to the combined markdown file

    Returns:
        Path to the created zip file
    """

    # Check if folder exists
    if not os.path.exists(folder_path):
        raise FileNotFoundError(f"Folder not found: {folder_path}")

    # Auto-generate zip filename if not provided
    if output_zip is None:
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        output_zip = f"extracted_pages_{timestamp}.zip"

    # Ensure .zip extension
    if not output_zip.endswith('.zip'):
        output_zip += '.zip'

    print(f"Creating zip file: {output_zip}")
    print(f"Source folder: {folder_path}")

    file_count = 0
    total_size = 0

    with zipfile.ZipFile(output_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
        # Add all files from the folder
        for root, dirs, files in os.walk(folder_path):
            for file in sorted(files):
                file_path = os.path.join(root, file)
                # Create relative path for the zip archive
                arcname = os.path.relpath(file_path, os.path.dirname(folder_path))

                zipf.write(file_path, arcname)
                file_size = os.path.getsize(file_path)
                total_size += file_size
                file_count += 1
                print(f"  ✓ Added: {arcname} ({file_size:,} bytes)")

        # Add combined file if it exists and requested
        if include_combined_file and os.path.exists(combined_file):
            zipf.write(combined_file, os.path.basename(combined_file))
            file_size = os.path.getsize(combined_file)
            total_size += file_size
            file_count += 1
            print(f"  ✓ Added: {os.path.basename(combined_file)} ({file_size:,} bytes)")

    zip_size = os.path.getsize(output_zip)
    compression_ratio = (1 - zip_size / total_size) * 100 if total_size > 0 else 0

    print(f"\n✓ Zip created successfully!")
    print(f"✓ Files added: {file_count}")
    print(f"✓ Original size: {total_size:,} bytes")
    print(f"✓ Compressed size: {zip_size:,} bytes")
    print(f"✓ Compression ratio: {compression_ratio:.1f}%")
    print(f"✓ Output: {output_zip}")

    return output_zip


def zip_with_custom_name(folder_path='extracted_pages', pdf_name='document'):
    """
    Create a zip file named after the PDF document

    Args:
        folder_path: Path to the folder containing extracted pages
        pdf_name: Name of the original PDF (without extension)

    Returns:
        Path to the created zip file
    """
    output_zip = f"{pdf_name}_extracted.zip"
    return zip_extracted_pages(folder_path, output_zip)


def zip_multiple_folders(folders, output_zip='all_extracted_pages.zip'):
    """
    Zip multiple extraction folders into one archive

    Args:
        folders: List of folder paths to include
        output_zip: Name of the output zip file

    Returns:
        Path to the created zip file
    """
    print(f"Creating zip file: {output_zip}")

    file_count = 0
    total_size = 0

    with zipfile.ZipFile(output_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for folder in folders:
            if not os.path.exists(folder):
                print(f"  ⚠ Skipping non-existent folder: {folder}")
                continue

            print(f"\nAdding folder: {folder}")

            for root, dirs, files in os.walk(folder):
                for file in sorted(files):
                    file_path = os.path.join(root, file)
                    # Keep folder structure in zip
                    arcname = os.path.relpath(file_path, os.path.dirname(folders[0]))

                    zipf.write(file_path, arcname)
                    file_size = os.path.getsize(file_path)
                    total_size += file_size
                    file_count += 1
                    print(f"  ✓ Added: {arcname}")

    zip_size = os.path.getsize(output_zip)

    print(f"\n✓ Zip created successfully!")
    print(f"✓ Files added: {file_count}")
    print(f"✓ Compressed size: {zip_size:,} bytes")
    print(f"✓ Output: {output_zip}")

    return output_zip


# =============================================================================
# GOOGLE COLAB SPECIFIC
# =============================================================================
def zip_and_download_colab(folder_path='extracted_pages', output_zip=None):
    """
    Create zip file and automatically download in Google Colab

    Args:
        folder_path: Path to the folder containing extracted pages
        output_zip: Name of output zip file (auto-generated if None)
    """
    try:
        from google.colab import files

        # Create the zip file
        zip_file = zip_extracted_pages(folder_path, output_zip)

        # Download it
        print(f"\nDownloading {zip_file}...")
        files.download(zip_file)
        print("✓ Download started!")

        return zip_file

    except ImportError:
        print("⚠ Not running in Google Colab. Zip file created but not auto-downloaded.")
        return zip_extracted_pages(folder_path, output_zip)


# =============================================================================
# MAIN EXECUTION
# =============================================================================
if __name__ == "__main__":
    import sys

    # Check if running in Colab
    try:
        from google.colab import files
        IN_COLAB = True
    except ImportError:
        IN_COLAB = False

    # Example usage based on environment
    if IN_COLAB:
        # In Google Colab - zip and auto-download
        print("Running in Google Colab")
        print("="*50)

        # Simple usage - zip extracted_pages folder
        zip_and_download_colab('extracted_pages')

    else:
        # Local execution - just create the zip
        print("Running locally")
        print("="*50)

        # Check command line arguments
        if len(sys.argv) > 1:
            folder = sys.argv[1]
            output = sys.argv[2] if len(sys.argv) > 2 else None
            zip_extracted_pages(folder, output)
        else:
            # Default - zip extracted_pages folder
            zip_extracted_pages('extracted_pages')

        print("\n" + "="*50)
        print("USAGE EXAMPLES:")
        print("="*50)
        print("1. Zip default folder:")
        print("   python zip_extracted_pages.py")
        print("\n2. Zip specific folder:")
        print("   python zip_extracted_pages.py my_pages")
        print("\n3. Zip with custom output name:")
        print("   python zip_extracted_pages.py extracted_pages my_output.zip")
        print("\n4. Use in Python script:")
        print("   from zip_extracted_pages import zip_extracted_pages")
        print("   zip_extracted_pages('extracted_pages', 'output.zip')")
```
